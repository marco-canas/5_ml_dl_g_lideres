El concepto de **muestra probabilística para poblaciones infinitas** es central en estadística y encaja muy bien con tu enfoque de modelación, simulación y pensamiento computacional.

---



# ¿Qué es una población infinita?

En términos de Estadística, una población infinita es:

* Un conjunto con **número ilimitado de elementos**, o
* Tan grande que **no se puede enumerar completamente**



### Ejemplos

* Resultados de lanzar una moneda indefinidamente
* Señales continuas en el tiempo
* Datos generados por un proceso estocástico

---



# ¿Qué es una muestra probabilística?

Es una muestra donde:

* Cada elemento tiene una **probabilidad conocida de ser seleccionado**
* Se usa el marco de la Teoría de la probabilidad

---



# 🔁 Caso: población infinita

Aquí no seleccionas individuos “uno por uno”, sino que:

👉 **Generas observaciones de una distribución**

Por ejemplo:

[
X \sim \mathcal{N}(\mu, \sigma)
]

Esto significa que tu “población infinita” está definida por una distribución.

---

# 📌 Idea clave

En lugar de decir:

> “tengo 1 millón de datos”

dices:

> “tengo una variable aleatoria con cierta distribución”

---

# 🔬 Fundamento matemático

La base está en el Teorema Central del Límite:

👉 Las medias muestrales tienden a una distribución normal, incluso si la población no lo es.

# 🧪 PRÁCTICA EN PYTHON (tipo clase UdeA)

Voy a proponerte una práctica completa, alineada con tu estilo:

---

## 🎯 Objetivo

Simular una **población infinita** y estudiar el comportamiento de las muestras.

---

## 📊 Paso 1: Definir la “población”

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# Población infinita simulada (distribución normal)
mu = 50
sigma = 10

# Generamos una "gran muestra" como aproximación
poblacion = np.random.normal(mu, sigma, 100000)
```

---

## 📈 Paso 2: Visualizar la población

```python
plt.figure()
plt.hist(poblacion, bins=50)
plt.title("Distribución de la población (aproximación)")
plt.show()
```

---

## 🎲 Paso 3: Tomar muestras probabilísticas

```python
def tomar_muestra(n):
    return np.random.normal(mu, sigma, n)

muestra = tomar_muestra(30)
print(muestra)
```

---

## 📊 Paso 4: Calcular estadísticos

```python
print("Media muestral:", np.mean(muestra))
print("Desviación estándar:", np.std(muestra))
```

---

## 🔁 Paso 5: Repetir muestreo (clave conceptual)

```python
medias = []

for _ in range(1000):
    muestra = tomar_muestra(30)
    medias.append(np.mean(muestra))
```

---

## 📉 Paso 6: Ver el Teorema Central del Límite

```python
plt.figure()
plt.hist(medias, bins=50)
plt.title("Distribución de medias muestrales")
plt.show()
```

---

# 🔍 Interpretación pedagógica

Aquí ocurre algo poderoso:

* La población puede ser cualquier distribución
* Pero las medias → se vuelven normales

💡 Esto es clave para:

* Inferencia estadística
* Machine Learning
* Modelos de predicción
