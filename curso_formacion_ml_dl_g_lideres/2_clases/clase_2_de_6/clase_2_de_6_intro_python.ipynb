#### Bloque 3: Ejercicios aplicados a administración (45 min)

### Ejercicio 1: Cálculo de ingresos

```python
precio = 25000
cantidad_vendida = 120

ingresos = precio * cantidad_vendida
print("Ingresos totales:", ingresos)
```

---

### Ejercicio 2: Utilidad de un producto

```python
costo_unitario = 18000
precio_venta = 30000
unidades = 200

utilidad_total = (precio_venta - costo_unitario) * unidades
print("Utilidad total:", utilidad_total)
```

---

### Ejercicio 3: Punto de equilibrio (simple)

```python
costos_fijos = 500000
precio = 20000
costo_variable = 12000

punto_equilibrio = costos_fijos / (precio - costo_variable)
print("Unidades para no tener pérdidas:", punto_equilibrio)
```

---

### Ejercicio 4: Presupuesto empresarial

```python
ventas = float(input("Ingrese ventas mensuales: "))
gastos = float(input("Ingrese gastos mensuales: "))

resultado = ventas - gastos

print("Resultado financiero:", resultado)
```
