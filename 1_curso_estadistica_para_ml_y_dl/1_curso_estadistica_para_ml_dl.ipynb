# Motivación al curso de estadística

A partir del Taller sobre Análisis Exploratorio de Datos y Preparación de datos para algoritmo de machine learning de regresión lineal y entrenamiento y medida de desempeño de un modelo lineal, (que incluye **EDA con `pandas`, visualización con `matplotlib`, `seaborn` y `plotly`, estandarización de datos, división `train_test_split`, regresión lineal y métricas como $R^2$**), el curso de **Estadística para Administración de Empresas – Programa G-Líderes** debe preparar a los estudiantes en tres niveles:

1. **Alfabetización en datos** (leer, interpretar y visualizar datos).
2. **Estadística descriptiva** (comprender estructuras de datos reales).
3. **Estadística inferencial y probabilidad** (fundamento conceptual de ML).

El objetivo es que **cuando lleguen a Machine Learning entiendan qué significan los datos, la variabilidad y la incertidumbre**, y no solo ejecuten código.

Dado que propones **encuentros quincenales de 1 hora**, propongo un curso de **20 encuentros (10 meses aproximadamente)**.

$$

$$

* moda

Interpretación empresarial.

Python:

```python
df.mean()
df.median()
df.mode()
```

## Encuentro 4

### Medidas de dispersión

Conceptos:

* rango
* varianza

$$
\sigma^2 = \frac{1}{n}\sum (x_i-\bar{x})^2
$$

* desviación estándar

$$
\sigma = \sqrt{\sigma^2}
$$

* coeficiente de variación

Interpretación empresarial.
