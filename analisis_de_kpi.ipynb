El **análisis de KPI (Key Performance Indicators)** es fundamental en ciencia de datos porque permite medir, monitorear y tomar decisiones basadas en el rendimiento de un sistema, negocio o proceso.

---



# ¿Qué es un KPI?

Un **KPI (Indicador Clave de Desempeño)** es una métrica cuantificable que responde a preguntas como:

* ¿Qué tan bien estamos funcionando?
* ¿Estamos cumpliendo objetivos?
* ¿Dónde debemos mejorar?



### Ejemplos típicos

* Ventas totales
* Tasa de conversión
* Retención de clientes
* Tiempo promedio de atención
* Error en predicciones (en modelos ML)

---

# 🧠 Enfoque matemático

Un KPI suele ser una función:

[
KPI = f(\text{variables del sistema})
]

Ejemplo:

[
\text{Tasa de conversión} = \frac{\text{clientes que compran}}{\text{clientes totales}}
]

---

# 🧪 Ejemplo práctico con Python

Voy a crear un **dataset simulado** tipo negocio (ventas online), y luego analizamos KPIs.

---

## 📊 1. Crear dataset

```python
import pandas as pd
import numpy as np

np.random.seed(42)

n = 200

df = pd.DataFrame({
    "fecha": pd.date_range(start="2024-01-01", periods=n, freq="D"),
    "visitas": np.random.randint(100, 1000, n),
    "compras": np.random.randint(10, 200, n),
    "ingresos": np.random.randint(500, 5000, n),
    "costos": np.random.randint(200, 3000, n)
})

df.head()
```

---

## 📈 2. Definir KPIs

### ✔️ KPI 1: Tasa de conversión

```python
df["conversion"] = df["compras"] / df["visitas"]
```

---

### ✔️ KPI 2: Ganancia

```python
df["ganancia"] = df["ingresos"] - df["costos"]
```

---

### ✔️ KPI 3: ROI (Return on Investment)

```python
df["ROI"] = df["ganancia"] / df["costos"]
```

---

## 📉 3. Análisis descriptivo

```python
df[["conversion", "ganancia", "ROI"]].describe()
```

Esto te da:

* media
* desviación estándar
* percentiles

---

## 📊 4. Visualización

```python
import matplotlib.pyplot as plt

plt.figure()
plt.plot(df["fecha"], df["conversion"])
plt.title("Tasa de conversión en el tiempo")
plt.xticks(rotation=45)
plt.show()
```

---

## 🔍 5. Interpretación

Ejemplo de lectura:

* 📈 Si la conversión sube → marketing efectivo
* 📉 Si el ROI baja → costos altos
* 💰 Si ganancia fluctúa → inestabilidad del negocio
